## **yfinance Library**
The **yfinance** is a popular open source python library that actually wraps a finance API for downloading stock market and financial data from Yahoo Finance. So whilst you are getting data from an open API, you don't need to write code to request data from the API and can just structure per the library outline to get your data. It works with the `pandas` library, returning data in a pandas dataframe object.

In this example, we source data from a number of companies within a selected start and end date.

You will notice we made it a function but really the `.download` is where the magic happens.

Refer to the [yfinance documentation](https://pypi.org/project/yfinance/) for full details on how to use this library.

```python
def get_yf_data(start, end, companies):
    """
    Retrieve historical market data for specified companies from Yahoo Finance.

    Args:
        start (str): Start date in a format compatible with the pd.to_datetime top-level function.
        end (str): End date in a format compatible with the pd.to_datetime top-level function
        companies (list): List of company ticker symbols.

    Returns:
        pd.DataFrame: A pandas DataFrame containing historical market data for the specified companies
                      within the given date range.
    """
    start_date = pd.to_datetime(start)
    end_date = pd.to_datetime(end)
    data =  yf.download(companies, start=start_date, end=end_date)
    return data
```

```python
yf_data = get_yf_data(
    start='January 2, 2011',
    end='5th of July, 2022',
    companies=['SPY', 'MSFT', 'AAPL', 'TSLA', 'GOOG']
)

yf_data
```

## **Requests Library**

If you don't have a nice library that wraps an API you can use the standard "requests" library in Python. This library simplifies HTTP requests, streamlining data retrieval from APIs. Here we are making more standard web API requests that are seen in all web interactions.

Let's look at an example open weather API. https://open-meteo.com/en/docs/historical-weather-api

```python
import requests

def fetch_weather_data(latitude, longitude, start_date, end_date, hourly_params):
    base_url = "https://archive-api.open-meteo.com/v1/archive"
    parameters = {
        "latitude": latitude,
        "longitude": longitude,
        "start_date": start_date,
        "end_date": end_date,
        "hourly": hourly_params
    }

    response = requests.get(base_url, params=parameters)

    if response.status_code == 200:
        return response.json()
    else:
        print(f"Error {response.status_code}: {response.text}")
        return None
```

```python
# Coordinates for Sydney, Australia
latitude = -33.8688
longitude = 151.2093
start_date = "2023-10-01"
end_date = "2023-10-01"
hourly_params = "temperature_2m" # You can extend this to other parameters if needed

data = fetch_weather_data(latitude, longitude, start_date, end_date, hourly_params)
print(data)
```

Extend the work above to:

- Use the above function to loop through dates between some start and end date. Say all of September 2023.
- Get the data and make an average daily temperature
- Make a dataframe of dates and average daily temperature



```python
from datetime import datetime, timedelta

# Define the start and end dates for September 2023
start_date = "2023-09-01"
end_date = "2023-09-30"

# Convert the start and end dates to datetime objects for the loop
start_date_obj = datetime.strptime(start_date, "%Y-%m-%d")
end_date_obj = datetime.strptime(end_date, "%Y-%m-%d")

# Generate a list of dates for September 2023
all_dates_september = [start_date_obj + timedelta(days=x) for x in range((end_date_obj - start_date_obj).days + 1)]

results = []

# Loop through dates
for current_date_obj in all_dates_september:
    data = fetch_weather_data(latitude, longitude, start_date, current_date_obj.strftime("%Y-%m-%d"), hourly_params)

    if data:
        # Create the average temperatied
        avg_temp = sum(data['hourly']['temperature_2m']) / len(data['hourly']['temperature_2m'])
        results.append((current_date_obj.strftime("%Y-%m-%d"), avg_temp))

# Convert results to a DataFrame
df = pd.DataFrame(results, columns=["date", "average_temperature"])
print(df)
```

## **pandas**

Before we jump to scraping raw HTML, Pandas itself actually has a `read_html`, a top-level function that simplifies web scraping by directly converting HTML tables into a list of pandas DataFrames, which can then be indexed. This works particularly well with tables from websites such as wikipedia.

```python
def read_html_data(url, table_index) -> pd.DataFrame:
    """
    Scrape data from the specified URL by reading HTML tables.

    Args:
        url (str): The URL of the webpage containing the HTML tables to be scraped.
        table_index (int): The index of the HTML table to be extracted from the webpage.

    Returns:
        pd.DataFrame: A pandas DataFrame containing the data extracted from the specified HTML table.
    """
    # Workaround using ssl library to ensure that data retrieval does not error out
    ssl._create_default_https_context = ssl._create_unverified_context

    list_of_dfs = pd.read_html(url)
    data = list_of_dfs[table_index]
    return data
```

```python
nrl_data = read_html_data(
    url='https://en.wikipedia.org/wiki/List_of_National_Rugby_League_records',
    table_index=4  # suppose we wanted the 5th table (python indices start at 0)
)

nrl_data
```

## **Beautiful Soup 4**

Beautiful Soup 4 is a widely used web scraping tool for extracting structured data from web pages, offering a simple solution in navigating through HTML content (that goes beyond just HTML tables).

Here we briefly showcase the library's power in a few lines of code, though you can easily add layers of complexity using loops or other techniques.

This script demonstrates how to extract t-shirt names and prices from a sports apparel website. Of course there are more machine learning specific datasets to create, but here we focus on the programatic aspect of collecting data.

```python
def scrape_product_prices(url: str='https://www.rebelsport.com.au/mens/clothing/tees-and-tops') -> pd.DataFrame:
    """
    Scrape product names and prices from the specified URL and return the data as a DataFrame.

    Args:
        url (str): The URL of the webpage

    Returns:
        pd.DataFrame: A pandas DataFrame containing product names and prices.
    """
    response = requests.get(url)
    soup = BeautifulSoup(response.content, 'html.parser')

    product_data = []
    all_products = soup.findAll('div', class_='product-tile gtm-product-click')

    for product in all_products:
        product_info = {}
        product_info['name'] = product.find('div', class_='product-name').text
        product_info['price'] = product.find('span', class_='price-sales').text
        product_data.append(product_info)

    return pd.DataFrame(product_data)

```

Call the function...

```python
tshirts = scrape_product_prices(
    url='https://www.rebelsport.com.au/mens/clothing/tees-and-tops'
)

tshirts
```

So what's happening here?

  
* **Web Request**:
   - Using the `requests.get(url)` method, the function fetches the HTML content of the given webpage. This is raw HTML data which can be very messy to look at. Try going to the website and `view source` to see what we are dealing with!

* **HTML Parsing**:
   - The fetched HTML content is then parsed using `BeautifulSoup` to create a soup object. This soup object allows for easy searching and extraction of specific elements from the HTML.

*  **Finding Products**:
   - The method `soup.findAll('div', class_='product-tile gtm-product-click')` searches for all `<div>` tags with the CSS class `product-tile gtm-product-click`.
   - Each matching element is considered a distinct product.
   - This is where knowledge of CSS selectors is needed. This is a large topic itself but there are many resources out there to help. Sometimes it is useful to inspect the source of a website initially to know what you are looking to scrape for. This can be done in the devtools (demonstration here).
   - Read more on CSS selectors and scraping [here](https://www.scrapingbee.com/blog/using-css-selectors-for-web-scraping/)
   
*  **Extracting Product Information**:
   - For each product, the function further finds and extracts the product name and price using the `product.find(...)` method combined with specific CSS selectors.
   - CSS selectors target specific elements based on their type, class, ID, or other attributes. In this function, CSS classes (e.g., `product-name` and `price-sales`) are used as selectors to locate desired content.
   
*  **Return as DataFrame**:
   - The extracted product information is stored in a list of dictionaries (`product_data`), which is then converted to a pandas DataFrame before being returned.


As you can see, the data looks rather messy. Data cleaning is arguably the most important stage of the analytics and modelling process, because the insights you obtain are only as good as the data at hand. We will adress this (and more) throughout the next sections of the course.

### **Your Turn**
Using Python, requests, and Beautiful Soup, scrape book information from the website - https://books.toscrape.com.

**Requirements**

1. For each book, extract:
*   Book title
*   Price
*   Star rating
*   Availability Status
2. Store the extracted data in a list of dictionaries and print all the book data.


```python
#Provide your answer with clear documentation that demonstrate a proper understanding of how the code works.

from bs4 import BeautifulSoup
import pandas as pd
import requests

# Helper dictionary to convert text ratings to numbers (optional but helpful)
RATING_MAP = {"One": 1, "Two": 2, "Three": 3, "Four": 4, "Five": 5}


def scrape_books(
    url: str = "https://books.toscrape.com/",
) -> list[dict[str, str | int]]:
    """Scrapes book title, price, star rating, and availability from books.toscrape.com."""
    response = requests.get(url)
    response.encoding = "utf-8"

    soup = BeautifulSoup(response.text, "html.parser")
    book_data = []

    # Find all book containers on the page
    all_books = soup.find_all("article", class_="product_pod")

    for book in all_books:
        # 1. Extract Full Title (stored in the 'title' attribute of the <a> tag)
        title_tag = book.find("h3").find("a")
        title = title_tag["title"]

        # 2. Extract Price
        price = book.find("p", class_="price_color").text.strip()

        # 3. Extract Star Rating (class name contains the rating, e.g., ['star-rating', 'Three'])
        rating_classes = book.find("p", class_="star-rating")["class"]
        # Filter out 'star-rating' to get the word rating (e.g., 'Three')
        rating_word = [
            cls for cls in rating_classes if cls != "star-rating"
        ][0]
        rating_num = RATING_MAP.get(rating_word, rating_word)

        # 4. Extract Availability Status
        availability = (
            book.find("p", class_="instock availability").text.strip()
        )

        # Store details in a dictionary
        book_info = {
            "title": title,
            "price": price,
            "rating": rating_num,
            "availability": availability,
        }

        book_data.append(book_info)

    return book_data


# Run scraper and print results
books = scrape_books()

# Print as list of dictionaries
for book in books:
    print(book)

# Optional: Convert to Pandas DataFrame for a clean visual representation
df = pd.DataFrame(books)
print("\n--- DataFrame Summary ---")
print(df.head())
```
